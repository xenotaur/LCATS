# Extract text from local HTML

Run with LCATS installed. This example is entirely offline, makes no files, and
uses the tested gatherers.extractors library. For real collection downloads see
[the gathering guide](../docs/how-to/gettenberg-gather.md). The old Lovecraft repair
prototype is available in Git history; it depended on an absent module.


```python
import bs4
from lcats.gatherers import extractors
```

```python
html = """<html><body>
<div id="pg-start-separator">Header</div>
<h1>A local example</h1><p>The travelers waited beside the river.</p>
<div id="pg-end-separator">Footer</div>
</body></html>"""
soup = bs4.BeautifulSoup(html, "lxml")
text = extractors.extract_text_between_ids(soup)
print(text)
```

For saved HTML, replace the authored string with `Path(path).read_text(encoding="utf-8")`.
Specify boundary IDs appropriate to that document; do not assume every HTML source
has Gutenberg separators. No metadata catalog is required.
